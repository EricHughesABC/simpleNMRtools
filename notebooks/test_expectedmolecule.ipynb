```python
mol = solution.expected_molecule
molprops = mol.molprops_df
```

```python
molgraph = nx.Graph()
atomNumbers = mol.molprops_df['atomNumber']

for atom in mol.GetAtoms():
    molgraph.add_node(
        atom.GetIdx(),
        symbol=atom.GetSymbol(),
        atom_number=atom.GetAtomicNum(),
        numProtons=atom.GetNumImplicitHs(),
        id0=atom.GetIdx(),
        x=mol.xy3_allatoms[atom.GetIdx()][0],
        y=mol.xy3_allatoms[atom.GetIdx()][1],
        ppm=-10000.0,
        atomNumber = allAtomsInfo[str(atom.GetIdx())]['atomNumber'],
    )

# Add edges for each bond in the molecule
for bond in mol.GetBonds():
    molgraph.add_edge(bond.GetBeginAtomIdx(), bond.GetEndAtomIdx())
```

```python
# import networkx as nx
from networkx.algorithms.approximation.steinertree import steiner_tree

# Make sure edges have a weight attribute (unweighted = 1 for all bonds)
for u, v in molgraph.edges():
    molgraph[u][v]['weight'] = 1

# Terminal nodes: one representative atom per symmetry-equivalent class
terminal_nodes = molprops['atom_idx'].tolist()

# Compute the Steiner tree
T = steiner_tree(molgraph, terminal_nodes, weight='weight')

# T is a subgraph (nx.Graph) — its nodes are the minimal atom set
# needed to connect all representative carbons
result_atoms = list(T.nodes())
print(f"Terminal (representative) atoms: {len(terminal_nodes)}")
print(f"Total atoms in minimal connected subgraph: {len(result_atoms)}")

# Optionally, split out which are your representative carbons vs. "connector" atoms
connector_atoms = set(result_atoms) - set(terminal_nodes)
print(f"Connector atoms needed: {connector_atoms}")
```
